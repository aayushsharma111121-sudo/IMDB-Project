# 24. Top 10 Comedy movies

```python
# Find movies containing Comedy in their genre

comedy_movies = data[data['Genre'].str.contains('Comedy', case=False, na=False)]

# Sort by rating and select top 10

top_10_comedy = comedy_movies.sort_values(
    by='Rating',
    ascending=False
).head(10)

top_10_comedy[['Title', 'Rating', 'Genre', 'Director']]
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top_10_comedy['Title'],
    top_10_comedy['Rating']
)

plt.xlabel('IMDb Rating')
plt.ylabel('Movie')
plt.title('Top 10 Highest Rated Comedy Movies')

plt.gca().invert_yaxis()
plt.show()
```

# 25. Top 10 movies in each genre

```python
# Split genres because some movies belong to multiple genres

genre_data = data.assign(
    Genre=data['Genre'].str.split(',')
).explode('Genre')

genre_data['Genre'] = genre_data['Genre'].str.strip()

# Display top 10 movies for each genre

for genre in genre_data['Genre'].unique():

    print("\n==============================")
    print("Genre:", genre)
    print("==============================")

    top_movies = genre_data[
        genre_data['Genre'] == genre
    ].sort_values(
        by='Rating',
        ascending=False
    ).head(10)

    print(top_movies[['Title', 'Rating']].to_string(index=False))
```

# 26. Most popular genres

```python
genre_counts = genre_data['Genre'].value_counts()

print(genre_counts)
```

```python
plt.figure(figsize=(10, 6))

genre_counts.head(10).plot(kind='bar')

plt.xlabel('Genre')
plt.ylabel('Number of Movies')
plt.title('Top 10 Most Common Movie Genres')

plt.xticks(rotation=45)
plt.show()
```

# 27. Average rating by genre

```python
average_rating_by_genre = genre_data.groupby('Genre')['Rating'].mean()

average_rating_by_genre = average_rating_by_genre.sort_values(
    ascending=False
)

print(average_rating_by_genre)
```

```python
plt.figure(figsize=(10, 6))

average_rating_by_genre.head(10).plot(kind='bar')

plt.xlabel('Genre')
plt.ylabel('Average IMDb Rating')
plt.title('Average IMDb Rating by Genre')

plt.xticks(rotation=45)
plt.show()
```

# 29. Highest-rated directors

```python
director_stats = data.groupby('Director').agg(
    Average_Rating=('Rating', 'mean'),
    Movie_Count=('Title', 'count')
)

# Consider directors with at least 3 movies
qualified_directors = director_stats[
    director_stats['Movie_Count'] >= 3
]

top_directors = qualified_directors.sort_values(
    by='Average_Rating',
    ascending=False
).head(10)

print(top_directors)
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top_directors.index,
    top_directors['Average_Rating']
)

plt.xlabel('Average IMDb Rating')
plt.ylabel('Director')
plt.title('Top Directors by Average IMDb Rating')

plt.gca().invert_yaxis()
plt.show()
```
